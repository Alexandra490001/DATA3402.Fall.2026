*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
def game_status(board):
    n = len(board)

    # Check rows
    for row in board:
        if row[0] != 0 and all(cell == row[0] for cell in row):
            return row[0]

    # Check columns
    for col in range(n):
        if board[0][col] != 0:
            if all(board[row][col] == board[0][col] for row in range(n)):
                return board[0][col]

    # Check main diagonal
    if board[0][0] != 0:
        if all(board[i][i] == board[0][0] for i in range(n)):
            return board[0][0]

    # Check other diagonal
    if board[0][n-1] != 0:
        if all(board[i][n-1-i] == board[0][n-1] for i in range(n)):
            return board[0][n-1]

    # Check for empty spaces
    for row in board:
        if 0 in row:
            return -1

    # Board is full and nobody won
    return 0
````

````python
# Test your solution here

winner_is_2 = [
    [1, 2, 2],
    [1, 2, 0],
    [2, 1, 1]
]

print(game_status(winner_is_2))
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
# Write your solution here

def make_move(board, player, row, col):
    n = len(board)

    if row < 0 or row >= n or col < 0 or col >= n:
        return False

    if board[row][col] != 0:
        return False

    board[row][col] = player

    return True
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
# Write your solution here

def make_labeled_move(board, player, location):
    location = location.upper()

    if len(location) < 2:
        return False

    row_letter = location[0]
    column_number = location[1:]

    if not column_number.isdigit():
        return False

    row = ord(row_letter) - ord("A")
    col = int(column_number) - 1

    return make_move(board, player, row, col)
````

````python
# Test your solution here

board = create_board(3)

make_labeled_move(board, 1, "A1")
make_labeled_move(board, 2, "B2")

draw_labeled_board(board)
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
# Write yourrr solution here

def play_game(n):
    board = create_board(n)
    player = 1

    while True:
        draw_labeled_board(board)

        player_move(board, player)

        status = game_status(board)

        if status == 1:
            draw_labeled_board(board)
            print("Player 1 won (X)!")
            break

        if status == 2:
            draw_labeled_board(board)
            print("Player 2 won (X)!")
            break

        if status == 0:
            draw_labeled_board(board)
            print("Draw!")
            break

        if player == 1:
            player = 2
        else:
            player = 1
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here

def computer_move(board, computer, opponent):
    n = len(board)

    # Check if computer can win
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = computer

                if game_status(board) == computer:
                    return

                board[row][col] = 0

    # Check if opponent can win, then block
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = opponent

                if game_status(board) == opponent:
                    board[row][col] = computer
                    return

                board[row][col] = 0

    # If nobody can win, take the first empty spot
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = computer
                return
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write your solution here

import copy

def minimax(board, player, computer, opponent, depth, max_depth):

    status = game_status(board)

    if status == computer:
        return 1

    if status == opponent:
        return -1

    if status == 0:
        return 0

    if depth == max_depth:
        return 0

    if player == computer:
        best_score = -10

        for i in range(len(board)):
            for j in range(len(board)):
                if board[i][j] == 0:

                    new_board = copy.deepcopy(board)
                    new_board[i][j] = computer

                    score = minimax(
                        new_board,
                        opponent,
                        computer,
                        opponent,
                        depth + 1,
                        max_depth
                    )

                    if score > best_score:
                        best_score = score

        return best_score

    else:
        best_score = 10

        for i in range(len(board)):
            for j in range(len(board)):
                if board[i][j] == 0:

                    new_board = copy.deepcopy(board)
                    new_board[i][j] = opponent

                    score = minimax(
                        new_board,
                        computer,
                        computer,
                        opponent,
                        depth + 1,
                        max_depth
                    )

                    if score < best_score:
                        best_score = score

        return best_score


def smart_computer_move(board, computer, opponent, max_depth):

    best_score = -10
    best_row = 0
    best_col = 0

    for i in range(len(board)):
        for j in range(len(board)):

            if board[i][j] == 0:

                new_board = copy.deepcopy(board)
                new_board[i][j] = computer

                score = minimax(
                    new_board,
                    opponent,
                    computer,
                    opponent,
                    1,
                    max_depth
                )

                if score > best_score:
                    best_score = score
                    best_row = i
                    best_col = j

    make_move(board, computer, best_row, best_col)
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Write your solution here

def computer_game(n, max_depth):
    board = create_board(n)
    player = 1

    while True:

        if player == 1:
            computer_move(board, 1)
        else:
            smart_computer_move(board, 2, 1, max_depth)

        status = game_status(board)

        if status != -1:
            return status

        if player == 1:
            player = 2
        else:
            player = 1


def run_games(n, max_depth):
    smart_wins = 0
    draws = 0

    for i in range(10):
        result = computer_game(n, max_depth)

        if result == 2:
            smart_wins += 1
        elif result == 0:
            draws += 1

    print("Board size:", n)
    print("Smart computer wins:", smart_wins)
    print("Draws:", draws)
    print("Smart computer win rate:", smart_wins / 10)
````

````python
# Test your solution here
run_games(3, 5)
run_games(4, 3)
run_games(5, 2)
````
