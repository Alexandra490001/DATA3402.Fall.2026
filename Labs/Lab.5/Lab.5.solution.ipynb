1. Create a `matrix` class with the following properties:
    * It can be initialized in 2 ways:
        1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    * Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
    * Matrix assignment works in 2 ways:
        1. If `M_1` and `M_2` are `matrix` instances `M_1=M_2` sets the values of `M_1` to those of `M_2`, if they are the same size. Error otherwise.
        2. In example above `M_2` can be a list of lists of correct size.


```python

class matrix:

    def __init__(self, n, m=None):

        if m is not None:

            self.data = []

            for i in range(n):

                row = []

                for j in range(m):
                    row.append(0)

                self.data.append(row)

        else:

            self.data = n

    def __getitem__(self, index):

        if isinstance(index, tuple):

            i = index[0]
            j = index[1]

            return self.data[i][j]

        else:

            return self.data[index]


    def __setitem__(self, index, value):

        if isinstance(index, tuple):

            i = index[0]
            j = index[1]

            self.data[i][j] = value

        else:

            self.data[index] = value


    def __repr__(self):

        return str(self.data)
```

2. Add the following methods:
    * `shape()`: returns a tuple `(n,m)` of the shape of the matrix.
    * `transpose()`: returns a new matrix instance which is the transpose of the matrix.
    * `row(n)` and `column(n)`: that return the nth row or column of the matrix M as a new appropriately shaped matrix object.
    * `to_list()`: which returns the matrix as a list of lists.
    *  `block(n_0,n_1,m_0,m_1)` that returns a smaller matrix located at the n_0 to n_1 columns and m_0 to m_1 rows. 
    * Modify `__getitem__` implemented above to support slicing.
        

```python
class matrix:

    def __init__(self, n, m=None):

        if m is not None:

            self.data = []

            for i in range(n):

                row = []

                for j in range(m):
                    row.append(0)

                self.data.append(row)

        else:

            self.data = n


    def __getitem__(self, index):

        if isinstance(index, tuple):

            i = index[0]
            j = index[1]

            if isinstance(i, int) and isinstance(j, int):
                return self.data[i][j]

            new_data = []

            for row in self.data[i]:
                new_row = []

                for value in row[j]:
                    new_row.append(value)

                new_data.append(new_row)

            return matrix(new_data)

        return self.data[index]


    def __setitem__(self, index, value):

        if isinstance(index, tuple):

            i = index[0]
            j = index[1]

            self.data[i][j] = value

        else:

            self.data[index] = value

    def shape(self):

        rows = len(self.data)
        columns = len(self.data[0])

        return (rows, columns)

    def transpose(self):

        rows = len(self.data)
        columns = len(self.data[0])

        new_data = []

        for j in range(columns):

            new_row = []

            for i in range(rows):
                new_row.append(self.data[i][j])

            new_data.append(new_row)

        return matrix(new_data)


    def row(self, n):

        return matrix([self.data[n]])

    def column(self, n):

        new_data = []

        for i in range(len(self.data)):
            new_data.append([self.data[i][n]])

        return matrix(new_data)

    def to_list(self):

        return self.data

    def block(self, n_0, n_1, m_0, m_1):

        new_data = []

        for i in range(n_0, n_1):

            new_row = []

            for j in range(m_0, m_1):
                new_row.append(self.data[i][j])

            new_data.append(new_row)

        return matrix(new_data)

    def __repr__(self):

        return str(self.data)

```

```python
A = matrix([
    [2, 4, 6],
    [1, 3, 8]
])

print("A =", A)

print("Shape =", A.shape())

print("Transpose =", A.transpose())

print("Row 0 =", A.row(0))

print("Column 1 =", A.column(1))

print("List =", A.to_list())

print("Block =", A.block(0, 2, 1, 3))

print("Slice =", A[0:2, 1:3])
```

3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python

def constant(n, m, c):

    data = []

    for i in range(n):

        row = []

        for j in range(m):
            row.append(float(c))

        data.append(row)

    return matrix(data)

def zeros(n, m):

    return constant(n, m, 0)



def ones(n, m):

    return constant(n, m, 1)



def eye(n):

    data = []

    for i in range(n):

        row = []

        for j in range(n):

            if i == j:
                row.append(1.0)
            else:
                row.append(0.0)

        data.append(row)

    return matrix(data)
```

```python

A = constant(2, 4, 5)
print("Constant:")
print(A)

B = zeros(2, 3)
print("Zeros:")
print(B)

C = ones(2, 3)
print("Ones:")
print(C)

D = eye(3)
print("Identity:")
print(D)
```

4. Add the following member functions to your class. Make sure to appropriately test the dimensions of the matrices to make sure the operations are correct.
    * `M.scalarmul(c)`: a matrix that is scalar product $cM$, where every element of $M$ is multiplied by $c$.
    * `M.add(N)`: adds two matrices $M$ and $N$. Don’t forget to test that the sizes of the matrices are compatible for this and all other operations.
    * `M.sub(N)`: subtracts two matrices $M$ and $N$.
    * `M.mat_mult(N)`: returns a matrix that is the matrix product of two matrices $M$ and $N$.
    * `M.element_mult(N)`: returns a matrix that is the element-wise product of two matrices $M$ and $N$.
    * `M.equals(N)`: returns true/false if $M==N$.

```python

class matrix:

    def __init__(self, n, m=None):

        if m is not None:
            self.data = []

            for i in range(n):
                row = []

                for j in range(m):
                    row.append(0)

                self.data.append(row)

        else:
            self.data = n


    def __getitem__(self, index):

        if isinstance(index, tuple):

            i = index[0]
            j = index[1]

            return self.data[i][j]

        return self.data[index]

    def __setitem__(self, index, value):

        if isinstance(index, tuple):

            i = index[0]
            j = index[1]

            self.data[i][j] = value

        else:
            self.data[index] = value

    def shape(self):

        return (len(self.data), len(self.data[0]))

    def transpose(self):

        new_data = []

        for j in range(len(self.data[0])):

            row = []

            for i in range(len(self.data)):
                row.append(self.data[i][j])

            new_data.append(row)

        return matrix(new_data)

    def row(self, n):

        return matrix([self.data[n]])

    def column(self, n):

        new_data = []

        for i in range(len(self.data)):
            new_data.append([self.data[i][n]])

        return matrix(new_data)
        
    def to_list(self):

        return self.data

    def block(self, n_0, n_1, m_0, m_1):

        new_data = []

        for i in range(n_0, n_1):

            row = []

            for j in range(m_0, m_1):
                row.append(self.data[i][j])

            new_data.append(row)

        return matrix(new_data)

    def scalarmul(self, c):

        new_data = []

        for i in range(len(self.data)):

            row = []

            for j in range(len(self.data[0])):
                row.append(self.data[i][j] * c)

            new_data.append(row)

        return matrix(new_data)

    def add(self, N):

        if self.shape() != N.shape():
            raise ValueError("Matrices must have the same size")

        new_data = []

        for i in range(len(self.data)):

            row = []

            for j in range(len(self.data[0])):
                row.append(self.data[i][j] + N.data[i][j])

            new_data.append(row)

        return matrix(new_data)

    def sub(self, N):

        if self.shape() != N.shape():
            raise ValueError("Matrices must have the same size")

        new_data = []

        for i in range(len(self.data)):

            row = []

            for j in range(len(self.data[0])):
                row.append(self.data[i][j] - N.data[i][j])

            new_data.append(row)

        return matrix(new_data)


    def mat_mult(self, N):

        if self.shape()[1] != N.shape()[0]:
            raise ValueError("Matrices cannot be multiplied")

        new_data = []

        for i in range(self.shape()[0]):

            row = []

            for j in range(N.shape()[1]):

                total = 0

                for k in range(self.shape()[1]):
                    total = total + self.data[i][k] * N.data[k][j]

                row.append(total)

            new_data.append(row)

        return matrix(new_data)

    def element_mult(self, N):

        if self.shape() != N.shape():
            raise ValueError("Matrices must have the same size")

        new_data = []

        for i in range(len(self.data)):

            row = []

            for j in range(len(self.data[0])):
                row.append(self.data[i][j] * N.data[i][j])

            new_data.append(row)

        return matrix(new_data)

    def equals(self, N):

        if self.shape() != N.shape():
            return False

        for i in range(len(self.data)):

            for j in range(len(self.data[0])):

                if self.data[i][j] != N.data[i][j]:
                    return False

        return True


    def __repr__(self):

        return str(self.data)

```

```python
A = matrix([
    [2, 4],
    [6, 8]
])

B = matrix([
    [1, 2],
    [3, 4]
])

print("Scalar:", A.scalarmul(2))
print("Add:", A.add(B))
print("Subtract:", A.sub(B))
print("Matrix multiply:", A.mat_mult(B))
print("Element multiply:", A.element_mult(B))
print("Equal:", A.equals(B))
```

5. Overload python operators to appropriately use your functions in 4 and allow expressions like:
    * 2*M
    * M*2
    * M+N
    * M-N
    * M*N
    * M==N


```python
class matrix(matrix):

    def __rmul__(self, c):
        return self.scalarmul(c)

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return self.scalarmul(other)
        else:
            return self.mat_mult(other)

    def __add__(self, N):
        return self.add(N)

    def __sub__(self, N):
        return self.sub(N)

    def __eq__(self, N):
        return self.equals(N)
```

```python
A = matrix([
    [2, 4],
    [6, 8]
])

B = matrix([
    [1, 2],
    [3, 4]
])

print("2 * A:")
print(2 * A)

print("A * 2:")
print(A * 2)

print("A + B:")
print(A + B)

print("A - B:")
print(A - B)

print("A * B:")
print(A * B)

print("A == B:")
print(A == B)
```
